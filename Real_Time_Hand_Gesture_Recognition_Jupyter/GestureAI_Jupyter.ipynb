```python
import os, subprocess, sys, time, urllib.request
from pathlib import Path

PROJECT_DIR = Path.cwd()
print("Project folder:", PROJECT_DIR)
print("Starting Flask server on http://127.0.0.1:5000 ...")

# Launch app.py as a child process so the notebook cell returns.
# Logs are written to flask_server.log in this folder.
log_file = open(PROJECT_DIR / "flask_server.log", "a", encoding="utf-8")
server_process = subprocess.Popen(
    [sys.executable, "app.py"],
    cwd=str(PROJECT_DIR),
    stdout=log_file,
    stderr=subprocess.STDOUT
)
time.sleep(3)
try:
    print(urllib.request.urlopen("http://127.0.0.1:5000/health", timeout=3).read().decode())
except Exception:
    print("Server is starting or dependencies/model are not ready yet. Check flask_server.log.")
print("Open: http://127.0.0.1:5000")
```

```python
# Optional: inspect the server log if the site does not load
from pathlib import Path
log_path = Path("flask_server.log")
print(log_path.read_text(encoding="utf-8", errors="replace")[-5000:] if log_path.exists() else "No log file yet.")
```

```python
# Run this cell to stop the web server process started above.
try:
    server_process.terminate()
    server_process.wait(timeout=5)
    print("GestureAI server stopped.")
except NameError:
    print("No server_process variable found in this kernel.")
except Exception as exc:
    print("Could not stop gracefully:", exc)
```
